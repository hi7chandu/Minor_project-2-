```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
display(df.head())

print("\nNumerical Columns:")
print(df.select_dtypes(include=np.number).columns.tolist())

print("\nCategorical Columns:")
print(df.select_dtypes(include="object").columns.tolist())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nDataset Information:")
df.info()
```

```python
before = len(df)

# Remove duplicate rows
df = df.drop_duplicates()

# Fill missing numerical values with median
num_cols = df.select_dtypes(include=np.number).columns

for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill missing categorical values with mode
cat_cols = df.select_dtypes(include="object").columns

for col in cat_cols:
    if df[col].notna().any():
        df[col] = df[col].fillna(df[col].mode()[0])

after = len(df)

print("Rows before cleaning:", before)
print("Rows after cleaning:", after)
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
print("Numerical Statistics:")
display(df.describe())

print("\nCategorical Statistics:")
display(df.describe(include="object"))
```

```python
if "price" in df.columns:
    sns.histplot(df["price"], kde=True)
    plt.title("Distribution of Car Prices")
    plt.xlabel("Price")
    plt.ylabel("Frequency")
    plt.show()
```

```python
if "year" in df.columns and "price" in df.columns:
    current_year = 2026

    df["car_age"] = current_year - df["year"]

    sns.scatterplot(
        data=df, x="car_age", y="price"
    )

    plt.title("Vehicle Age vs Car Price")
    plt.xlabel("Car Age")
    plt.ylabel("Price")
    plt.show()
```

```python
if "mileage" in df.columns and "price" in df.columns:
    sns.scatterplot(
        data=df, x="mileage", y="price"
    )

    plt.title("Mileage vs Car Price")
    plt.xlabel("Mileage")
    plt.ylabel("Price")
    plt.show()
```

```python
if "transmission" in df.columns and "price" in df.columns:
    sns.boxplot(
        data=df, x="transmission", y="price"
    )

    plt.title("Transmission Type vs Car Price")
    plt.xlabel("Transmission")
    plt.ylabel("Price")
    plt.show()

    print("Average Price by Transmission:")
    print(df.groupby("transmission")["price"].mean())
```

```python
if "brand" in df.columns and "price" in df.columns:
    avg_price = (
        df.groupby("brand")["price"]
        .mean()
        .sort_values(ascending=False)
    )

    print("Top 10 Brands by Average Price:")
    print(avg_price.head(10))

    avg_price.head(10).plot(
        kind="bar", figsize=(8, 4)
    )

    plt.title("Top 10 Brands by Average Car Price")
    plt.xlabel("Brand")
    plt.ylabel("Average Price")
    plt.xticks(rotation=45)
    plt.show()
```

```python
if "price" in df.columns:
    sns.boxplot(x=df["price"])

    plt.title("Outliers in Car Prices")
    plt.xlabel("Price")
    plt.show()

    Q1 = df["price"].quantile(0.25)
    Q3 = df["price"].quantile(0.75)

    IQR = Q3 - Q1

    outliers = df[
        (df["price"] < Q1 - 1.5 * IQR) |
        (df["price"] > Q3 + 1.5 * IQR)
    ]

    print("Number of Price Outliers:", len(outliers))
    display(outliers.head())
```

```python
# Find columns matching common car dataset names
columns = {c.lower().strip(): c for c in df.columns}

price_col = next(
    (columns[c] for c in ["price", "selling_price", "selling price", "price_inr"]
     if c in columns), None
)

mileage_col = next(
    (columns[c] for c in ["mileage", "km_driven", "kms_driven"]
     if c in columns), None
)

year_col = next(
    (columns[c] for c in ["year", "model_year"]
     if c in columns), None
)

if price_col:
    sns.histplot(df[price_col], kde=True)
    plt.title("Distribution of Car Prices")
    plt.show()

if mileage_col and price_col:
    sns.scatterplot(data=df, x=mileage_col, y=price_col)
    plt.title("Mileage vs Car Price")
    plt.show()

if year_col and price_col:
    sns.scatterplot(data=df, x=year_col, y=price_col)
    plt.title("Year vs Car Price")
    plt.show()

if price_col:
    print("Average Car Price:", df[price_col].mean())
```

```python
# Numerical variable distributions
for col in num_cols[:3]:
    sns.histplot(data=df, x=col, kde=True)
    plt.title("Distribution of " + col)
    plt.show()

# Most common categories
for col in cat_cols[:2]:
    df[col].value_counts().head(10).plot(kind="bar")
    plt.title("Top Categories in " + col)
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
# Relationship between numerical variables
if len(num_cols) >= 2:
    sns.scatterplot(
        data=df, x=num_cols[0], y=num_cols[1]
    )
    plt.title(num_cols[0] + " vs " + num_cols[1])
    plt.show()

# Numerical variable compared with a category
if len(num_cols) >= 1 and len(cat_cols) >= 1:
    top_categories = df[cat_cols[0]].value_counts().head(10).index
    temp = df[df[cat_cols[0]].isin(top_categories)]

    sns.boxplot(
        data=temp, x=cat_cols[0], y=num_cols[0]
    )
    plt.title(num_cols[0] + " by " + cat_cols[0])
    plt.xticks(rotation=45)
    plt.show()
```

```python
if len(num_cols) >= 2:
    plt.figure(figsize=(10, 6))

    sns.heatmap(
        df[num_cols].corr(),
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )

    plt.title("Correlation Heatmap")
    plt.show()
```

```python
# Outlier detection using IQR
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    outliers = df[
        (df[col] < Q1 - 1.5 * IQR) |
        (df[col] > Q3 + 1.5 * IQR)
    ]

    print(col, "- Number of outliers:", len(outliers))

# Five data-based findings
print("\n===== KEY INSIGHTS =====")

if len(cat_cols) > 0:
    col = cat_cols[0]
    print("1. Most common", col, ":",
          df[col].mode()[0])

if len(num_cols) > 0:
    col = num_cols[0]
    print("2. Average", col, ":",
          round(df[col].mean(), 2))

    print("3. Highest", col, ":",
          df[col].max())

    print("4. Lowest", col, ":",
          df[col].min())

if len(num_cols) >= 2:
    corr = df[num_cols].corr()
    pairs = corr.where(
        np.triu(np.ones(corr.shape), k=1).astype(bool)
    ).stack()

    if not pairs.empty:
        pair = pairs.abs().idxmax()
        value = pairs.loc[pair]

        print("5. Strongest numerical relationship:",
              pair[0], "and", pair[1],
              "with correlation", round(value, 2))
```

```python
df['year'] = pd.to_numeric(df['year'], errors='coerce')
df['car_age'] = 2026 - df['year']

sns.scatterplot(
    data=df, x='car_age', y='price_inr'
)
plt.title("Car Age vs Price in INR")
plt.xlabel("Car Age (Years)")
plt.ylabel("Price in INR")
plt.show()

print("Correlation between age and price:",
      df['car_age'].corr(df['price_inr']))
```

```python
df['mileage_num'] = pd.to_numeric(
    df['mileage_kmpl'].astype(str).str.extract(
        r'([\d.]+)'
    )[0],
    errors='coerce'
)

sns.scatterplot(
    data=df, x='mileage_num', y='price_inr'
)
plt.title("Mileage vs Price in INR")
plt.xlabel("Mileage (kmpl)")
plt.ylabel("Price in INR")
plt.show()

print("Mileage-Price Correlation:",
      df['mileage_num'].corr(df['price_inr']))
```
